The following code provides an outline of the SQL schema used to create the Form N-PORT Holdings Database. To utilize this code you will first need to set up your own container for housing the data and input your own connection information.

```python
#create database
import psycopg2

def create_tables(conn):
    try:
        # create a cursor
        cur = conn.cursor()

        # create tables
        create_table_commands = (
            """
            CREATE TABLE nport_filings (
                cik VARCHAR(20),
                series_id VARCHAR(20),
                Registrant_Name VARCHAR(255),
                Period_Of_Report DATE,
                Filing_Date DATE
            )
            """,
            """
            CREATE TABLE nport_holding_infos (
                isin VARCHAR(20),
                lei VARCHAR(50),
                title VARCHAR(255),
                cusip VARCHAR(20),
                issuerCat VARCHAR(50)
            )
            """,
            """
            CREATE TABLE nport_holdings (
                series_id VARCHAR(20),
                Period_Of_Report DATE,
                Filing_Date DATE,
                isin VARCHAR(20),
                name VARCHAR(255),
                balance FLOAT,
                units VARCHAR(50),
                curCd VARCHAR(10),
                valUSD FLOAT,
                pctVal FLOAT,
                payoffprofile VARCHAR(10)
            )
            """
        )

        # create table one by one
        for command in create_table_commands:
            cur.execute(command)
        
        # close cursor
        cur.close()
        
        # make the changes to the database persistent
        conn.commit()
        print("Tables created successfully.")
    except Exception as error:
        print(f"Error: {error}")
        conn.rollback()
    finally:
        if cur is not None:
            cur.close()

# Usage example
try:
    conn = psycopg2.connect(
    host="localhost",
    port='5435',
    database="nport",
    user="example",
    password="example")

    create_tables(conn)

finally:
    # Close the connection after use
    if conn is not None:
        conn.close()
```
